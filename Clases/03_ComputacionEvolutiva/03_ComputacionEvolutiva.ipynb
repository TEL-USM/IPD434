# Resolviendo problemas: Método de la Caja Negra

Es un método que sirve para clasificar los diversos tipos de problemas que necesitan ser resueltos en la vida real. Se compone de tres elementos:
* Una entrada, que representa la información de entrada a del sistema;
* Un modelo, que toma al entrada y la utiliza, transforma y/o modifica; y
* Una salida, que representa el resultado generado por el modelo.

# Optimización Combinatoria

Campo de la matemática que busca encontrar el objeto optimo que resuelve a un problema dentro de un conjunto finito de objetos.  

## Problemas de Optimización Combinatoria

* Problemas de Optmizaición
    * Modelo Matemáticos (model) + Función de Evaluación (output). Buscar valor de variables (input)
* Problemas de Modelado
    * Data set (input) + data labeling/prediction (output). Sistema aprendido (model), se transforma en optimización (Minimizar error cuadrático medio, maximizar éxíto).
    

* The candidates for solving these kind of problems are located in a space of possibilities which is usually enormous.
* Then, the problem solving process is a search through a potentially huge set of possibilities to find a solution.
    * **Search State Space**.
* Thus, combinatorial optimisation problems can be seen as search problems.

## Búsqueda

* To find, evaluate and accept solutions as candidates to solve a problem from the set of all possible solutions.

#### To find:
  * **Search Space** $\rightarrow$ the set of all possible solutions.
  * **Search Space** $\rightarrow$ domains of all decision variables that define a problem solution.

#### To evaluate:
   * Using a specific function called **evaluation** or **objective function** that quantifies and compares candidates.

#### To accept:
  * If a **set of constraints** is fulfilled.
