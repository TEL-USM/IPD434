```python
cr.create("Fitness", bs.Fitness, weights=(-1.0,)) #Creamos Clases de Función Fitness
cr.create("Individuo", list, fitness=cr.Fitness) #Creamos Clase de Individuo
```

```python
board_size = 6

toolbox = bs.Toolbox()
toolbox.register("fill_int", rd.randint, 1, board_size)
toolbox.register("fill_ind", tl.initRepeat, cr.Individuo, toolbox.fill_int, n=board_size)
```

```python
ind1 = toolbox.fill_ind()
print(ind1)
print(ind1.fitness.valid)
```
