# Soft Computing

- El término **Soft Computing** fue introducidor por Lotfi A. Zadeh.
- Soft Computing es un conjunto de técnicas capaces de resolver **problemas de alta complejidad** con técnicas **imprecisas o incompletas**.

# Ejemplos

#### Estacionar automóvil en un espacio estrecho

No se necesitan las dimensiones **exactas** del espacio o del automóvil.

#### Reconocer caracteres escritos a mano

Aunque las letras escritas a mano no sigan su forma ideal, igualmente **somos capaces de reconocerlos**.

#### Hormigas recolectando comida

Inicialmente las hormigas buscan comida aleatoriamente. Cuando  una de ellas encuentra comida, **comparte el mensaje** con el resto para que todas se dirijan a la fuente de alimento.

# Conceptos importantes

## Espacio de Estados/Búsqueda

- Corresponde al conjunto de estados o configuraciones posibles para un sistema.
- La búsqueda sobre espacio de estados se realiza desde un **estado inicial** y sobre todas las transiciones de estado que permitan llegar a un estado objetivo o **estado final**. Este estado final se define en base al problema a resolver.

## P vs NP

Los problemas pueden ser clasificados en diferentes clases:

- Los **problemas P**, se les llama así pues pueden ser resueltos en tiempo polinomial. Dicho en otras palabras, existe una función polinómica que acota el tiempo de resolución del problema en base a la entrada. Este tipo de problemas son considerados de fácil resolución.
- Los **problemas NP** (de sus siglas en inglés Nondeterministic Polynomial time) son aquellos en los que, dada una solución, esta se puede verificar en tiempo polinomial. Los problemas NP no necesariamente han sido resueltos en tiempo polinomial.

El problema **P vs NP** intenta determinar si el conjunto **P** es igual al conjunto **NP**. Vale decir, **¿es todo problema que se pueda verificar en tiempo polinomial también ser resuelto en tiempo polinomial?** (aún sin respuesta

## NP-Complete

**NP-Completo** son aquellos problemas de tipo NP tales que cualquier problema NP puede ser traducido a este en tiempo polinomial.

## NP-Hard

Se les llama problemas **NP-Hard** a aquellos que, si bien no necesariamente se encuentran en el conjunto NP, pueden llegar a ser tan difíciles como los problemas más complejos de NP.
