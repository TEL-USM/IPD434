# Introducción

## Resultados de aprendizaje

- Analiza la complejidad de un problema y de su espacio de búsqueda de soluciones.
- Analiza los diferentes criterios de aprendizaje presente en los modelos de Soft Computing.
- Evalúa la aplicación de algoritmos de Soft Computing en distintos problemas complejos.

## Contenidos

- Introducción a Soft Computing.
- Sistemas difusos (Fuzzy Systems).
- Espacio de búsqueda y complejidad de problemas.
- Computación evolutiva y algoritmos bioinspirados.
- Redes Neuronales.
