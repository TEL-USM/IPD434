# Clasificación de texto con Transformers

Uso de una pipeline con modelo explícito y pruebas de sensibilidad.

## Dependencias

En un entorno nuevo: `%pip install transformers torch pandas`. La primera ejecución descarga pesos y tokenizer.

```python
import pandas as pd
from transformers import AutoTokenizer, pipeline

MODEL_ID = 'distilbert/distilbert-base-uncased-finetuned-sst-2-english'
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
classifier = pipeline('text-classification', model=MODEL_ID, tokenizer=tokenizer, device=-1)
```

```python
texts = [
    'The service was fast and reliable.',
    'The connection failed three times today.',
    'The interface is simple, but the documentation is incomplete.',
    'It works exactly as specified.',
]
results = classifier(texts, truncation=True)
pd.DataFrame([{**{'text': text}, **result} for text, result in zip(texts, results)])
```

## Inspección de tokenización

```python
sample = texts[2]
encoded = tokenizer(sample, truncation=True)
print(tokenizer.convert_ids_to_tokens(encoded['input_ids']))
print('Número de tokens:', len(encoded['input_ids']))
```

## Pruebas de sensibilidad

El modelo fue entrenado para sentimiento en inglés. Ironía, negación compleja, idioma y dominio técnico pueden producir errores confiados.

```python
stress_tests = [
    'Great, another outage. Exactly what we needed.',
    'The service is not bad.',
    'La conexión funciona correctamente.',
    'Packet loss decreased from 3 percent to 0.1 percent.',
]
pd.DataFrame([{**{'text': text}, **result} for text, result in zip(stress_tests, classifier(stress_tests, truncation=True))])
```
